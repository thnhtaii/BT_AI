```python
def in_mt(mt):
    for i in range(0, 9, 3):
        print(f'{mt[i]}  {mt[i+1]}  {mt[i+2]}')

def nhap_mt(ten):
    print(f'Nhập ma trận {ten}')
    mt = []
    for i in range(3):
        while True:
            hang = input(f'Nhập hàng {i+1}: ').split()
            if len(hang) == 3:
                for so in hang:
                    mt.append(int(so))
                break
            else:
                print('Nhập đúng 3 số')
    return mt

def giai_toan():
    trang_thai_dau = nhap_mt("BẮT ĐẦU")
    trang_thai_dich = nhap_mt("ĐÍCH")
    lich_su = [{'ban_co': trang_thai_dau, 'cha': -1, 'hanh_dong': 'Bắt đầu'}]
    danh_sach_cho = [0]
    da_tung_qua = [trang_thai_dau]

    while danh_sach_cho:
        chi_so_hien_tai = danh_sach_cho.pop(0)
        nut_hien_tai = lich_su[chi_so_hien_tai]

        if nut_hien_tai['ban_co'] == trang_thai_dich:
            print("\nChuỗi hành động là: ")
            duong_di = []
            hanh_dong_di = []
            truy_vet = chi_so_hien_tai
            while truy_vet != -1:
                duong_di.append(lich_su[truy_vet]['ban_co'])
                hanh_dong_di.append(lich_su[truy_vet]['hanh_dong'])
                truy_vet = lich_su[truy_vet]['cha']
            
            for buoc, (bc, hd) in enumerate(zip(reversed(duong_di), reversed(hanh_dong_di))):
                print(f"{buoc}: {hd}")
                in_mt(bc)
            return

        vi_tri_0 = nut_hien_tai['ban_co'].index(0)
        hang = vi_tri_0 // 3
        cot = vi_tri_0 % 3

        if hang > 0:
            ban_co_moi = list(nut_hien_tai['ban_co'])
            ban_co_moi[vi_tri_0], ban_co_moi[vi_tri_0 - 3] = ban_co_moi[vi_tri_0 - 3], ban_co_moi[vi_tri_0]
            if ban_co_moi not in da_tung_qua:
                da_tung_qua.append(ban_co_moi)
                lich_su.append({'ban_co': ban_co_moi, 'cha': chi_so_hien_tai, 'hanh_dong': 'Lên'})
                danh_sach_cho.append(len(lich_su) - 1)

        if hang < 2:
            ban_co_moi = list(nut_hien_tai['ban_co'])
            ban_co_moi[vi_tri_0], ban_co_moi[vi_tri_0 + 3] = ban_co_moi[vi_tri_0 + 3], ban_co_moi[vi_tri_0]
            if ban_co_moi not in da_tung_qua:
                da_tung_qua.append(ban_co_moi)
                lich_su.append({'ban_co': ban_co_moi, 'cha': chi_so_hien_tai, 'hanh_dong': 'Xuống'})
                danh_sach_cho.append(len(lich_su) - 1)

        if cot > 0:
            ban_co_moi = list(nut_hien_tai['ban_co'])
            ban_co_moi[vi_tri_0], ban_co_moi[vi_tri_0 - 1] = ban_co_moi[vi_tri_0 - 1], ban_co_moi[vi_tri_0]
            if ban_co_moi not in da_tung_qua:
                da_tung_qua.append(ban_co_moi)
                lich_su.append({'ban_co': ban_co_moi, 'cha': chi_so_hien_tai, 'hanh_dong': 'Trái'})
                danh_sach_cho.append(len(lich_su) - 1)

        if cot < 2:
            ban_co_moi = list(nut_hien_tai['ban_co'])
            ban_co_moi[vi_tri_0], ban_co_moi[vi_tri_0 + 1] = ban_co_moi[vi_tri_0 + 1], ban_co_moi[vi_tri_0]
            if ban_co_moi not in da_tung_qua:
                da_tung_qua.append(ban_co_moi)
                lich_su.append({'ban_co': ban_co_moi, 'cha': chi_so_hien_tai, 'hanh_dong': 'Phải'})
                danh_sach_cho.append(len(lich_su) - 1)

    print("Không tìm thấy đường đi!")

if __name__ == "__main__":
    giai_toan()
```
