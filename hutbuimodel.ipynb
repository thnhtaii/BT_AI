```python
import random
import time
from IPython.display import clear_output

def in_mt(mt, r, c):
    for i in range(3):
        row = ""
        for j in range(3):
            val = mt[i*3 + j]
            row += f"[{val}] " if i == r and j == c else f" {val}  "
        print(row)
    print("-" * 20)

def di_chuyen(mt_cu, r_moi, c_moi, hanh_dong):
    return {
        'mt': list(mt_cu),
        'r': r_moi,
        'c': c_moi,
        'hd': hanh_dong
    }

def giai_ngau_nhien():
    # STATE
    mt = []
    print("Nhập ma trận sàn nhà 3x3: ")
    for i in range(3):
        while True:
            hang = input(f"Nhập hàng {i+1}: ").split()
            if len(hang) == 3:
                mt.extend([int(x) for x in hang])
                break
            else:
                print("Phải nhập đúng 3 số!")

    res_input = input("Vị trí robot (Hàng Cột): ").split()
    r, c = map(int, res_input)
    r, c = r - 1, c - 1
    buoc = 0
    hd_in = "Bắt đầu"
    da_di_qua = []

    while sum(mt) > 0:
        trang_thai_hien_tai = (tuple(mt), r, c)
        if trang_thai_hien_tai not in da_di_qua:
            da_di_qua.append(trang_thai_hien_tai)
        buoc += 1
        pos = r * 3 + c
        # RULES
        tap_luat_kha_thi = []

        if mt[pos] == 1:
            mt_sau_hut = list(mt)
            mt_sau_hut[pos] = 0
            if (tuple(mt_sau_hut), r, c) not in da_di_qua:
                tap_luat_kha_thi.append("Hut")
        if r > 0:
            if (tuple(mt), r - 1, c) not in da_di_qua:
                tap_luat_kha_thi.append("Len")
        if r < 2:
            if (tuple(mt), r + 1, c) not in da_di_qua:
                tap_luat_kha_thi.append("Xuong")
        if c > 0:
            if (tuple(mt), r, c - 1) not in da_di_qua:
                tap_luat_kha_thi.append("Trai")
        if c < 2:
            if (tuple(mt), r, c + 1) not in da_di_qua:
                tap_luat_kha_thi.append("Phai")

        # MODEL
        if not tap_luat_kha_thi:
            print("Robot bị kẹt: Mọi hướng đi tiếp theo đều đã từng đi qua!")
            break

        hanh_dong_chon = random.choice(tap_luat_kha_thi)

        # ACTION
        if hanh_dong_chon == "Hut":
            mt[pos] = 0
            hd_in = "Hút bụi"
        elif hanh_dong_chon == "Len":
            r -= 1
            hd_in = "Di chuyển Lên"
        elif hanh_dong_chon == "Xuong":
            r += 1
            hd_in = "Di chuyển Xuống"
        elif hanh_dong_chon == "Trai":
            c -= 1
            hd_in = "Di chuyển Sang Trái"
        elif hanh_dong_chon == "Phai":
            c += 1
            hd_in = "Di chuyển Sang Phải"

        clear_output(wait=True)
        print(f"Bước {buoc}: {hd_in}")
        in_mt(mt, r, c)
        time.sleep(0.5)

        if buoc > 100:
            print("Dừng lại!")
            break

    if sum(mt) == 0:
        print(f"Sàn nhà đã sạch sau {buoc} bước")

if __name__ == "__main__":
    giai_ngau_nhien()
```
