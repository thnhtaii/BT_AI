```python
import random
import time

def in_mt(mt):
    for i in range(3):
        row = ""
        for j in range(3):
            val = mt[i*3 + j]
            if val == 0:
                row += " [ ] "
            else:
                row += f"  {val}  "
        print(row)
    print("-" * 20)

def swap(mt, i, j):
    new_mt = list(mt)
    new_mt[i], new_mt[j] = new_mt[j], new_mt[i]
    return new_mt

def giai_ngau_nhien_8puzzle():
    mt = []
    dich = [1, 2, 3, 4, 5, 6, 7, 8, 0]

    print("Nhập ma trận 8-puzzle: ")
    for i in range(3):
        while True:
            hang = input(f"Nhập hàng {i+1}: ").split()
            if len(hang) == 3:
                mt.extend([int(x) for x in hang])
                break
            else:
                print("Nhập đúng 3 số!")

    buoc = 0
    hd_in = "Bắt đầu"
    da_di_qua = []

    print(f"\nBước {buoc}: {hd_in}")
    in_mt(mt)

    while mt != dich:
        trang_thai_hien_tai = tuple(mt)
        if trang_thai_hien_tai not in da_di_qua:
            da_di_qua.append(trang_thai_hien_tai)

        buoc += 1
        pos = mt.index(0)
        r, c = pos // 3, pos % 3

        tap_luat_kha_thi = []

        if r > 0:
            mt_moi = swap(mt, pos, pos - 3)
            if tuple(mt_moi) not in da_di_qua:
                tap_luat_kha_thi.append(("Lên", mt_moi))
        if r < 2:
            mt_moi = swap(mt, pos, pos + 3)
            if tuple(mt_moi) not in da_di_qua:
                tap_luat_kha_thi.append(("Xuống", mt_moi))
        if c > 0:
            mt_moi = swap(mt, pos, pos - 1)
            if tuple(mt_moi) not in da_di_qua:
                tap_luat_kha_thi.append(("Trái", mt_moi))
        if c < 2:
            mt_moi = swap(mt, pos, pos + 1)
            if tuple(mt_moi) not in da_di_qua:
                tap_luat_kha_thi.append(("Phải", mt_moi))

        if not tap_luat_kha_thi:
            print("Robot bị kẹt!")
            break

        hanh_dong_chon, mt = random.choice(tap_luat_kha_thi)
        hd_in = f"Di chuyển ô trống sang {hanh_dong_chon}"

        print(f"Bước {buoc}: {hd_in}")
        in_mt(mt)

        time.sleep(0.1)

        if buoc > 1000:
            print("Quá số bước quy định!")
            break

    if mt == dich:
        print(f"Thành công sau {buoc} bước")

if __name__ == "__main__":
    giai_ngau_nhien_8puzzle()
```
